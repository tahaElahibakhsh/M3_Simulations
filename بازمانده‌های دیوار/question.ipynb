## ۳. لجستیک‌رگرشن از صفر
با گرادیان نزولی و cross-entropy وزن‌ها و بایاس را یاد بگیرید. L2 را فقط روی وزن‌ها اعمال و احتمال‌ها را برای محاسبهٔ لگاریتم پایدار کنید.

</div>

```python
Xtr=train_df[FEATURES].to_numpy(float); Xv=validation_df[FEATURES].to_numpy(float)
ytr=train_df.label.to_numpy(int); yv=validation_df.label.to_numpy(int)
mu,sd=fit_standardizer(Xtr)
raw_model=fit_logistic_gd(apply_standardizer(Xtr,mu,sd),ytr)
raw_pred=predict_labels(raw_model,apply_standardizer(Xv,mu,sd))
print("Raw macro-F1:",f1_score(yv,raw_pred,average="macro"))
```

## ۴. خط مبنا: ویژگی‌های خام
مدل را با هفت ویژگی خام آموزش دهید و macro-F1 بخش validation را ثبت کنید.

</div>

```python
Xtr=build_features(train_df); Xv=build_features(validation_df)
mu,sd=fit_standardizer(Xtr); Xtr=apply_standardizer(Xtr,mu,sd); Xv=apply_standardizer(Xv,mu,sd)
model=fit_logistic_gd(Xtr,ytr,learning_rate=0.08,epochs=3000,l2=0.03)
val_prob=predict_proba(model,Xv); threshold=0.5
val_pred=(val_prob>=threshold).astype(int)
print("Engineered macro-F1:",f1_score(yv,val_pred,average="macro"))
print("Accuracy:",accuracy_score(yv,val_pred),"confusion:",confusion_matrix(yv,val_pred))
```

## ۵. آجرهای ویژگی
فشردگی 4πA/P²، نسبت محور کوچک به بزرگ، و توپُری A/ConvexArea را اضافه کنید. توضیح دهید هرکدام چه چیزی از شکل اندازه می‌گیرد. عنوان Another Brick in the Wall در این دفترچه استعارهٔ ساختن تدریجی سرنخ‌هاست؛ متن ترانه برای حل لازم نیست.

</div>

```python
full=pd.concat([train_df,validation_df],ignore_index=True)
Xfull=build_features(full); yfull=full.label.to_numpy(int); mu,sd=fit_standardizer(Xfull)
final_model=fit_logistic_gd(apply_standardizer(Xfull,mu,sd),yfull,learning_rate=0.08,epochs=3000,l2=0.03)
predictions=predict_labels(final_model,apply_standardizer(build_features(test_df),mu,sd),threshold)
submission=pd.DataFrame({"id":test_df.id,"prediction":predictions})
assert len(predictions)==len(test_df) and submission.id.is_unique
submission.to_csv("submission.csv",index=False)
```
