## مأموریت شما

نیروگاه سپهر برای جلوگیری از نوسان در شبکه باید توان خروجی توربین‌هایش را پیش‌بینی کند. حسگرها دما، فشار، سرعت چرخش و وضعیت کارکرد را ثبت می‌کنند؛ اما داده‌ها گاهی ناقص‌اند و قالب بعضی اندازه‌گیری‌ها یکسان نیست. از روی داده‌های ثبت‌شده، مقدار `power_output` را پیش‌بینی کنید.

کنترل‌گر نیروگاه فقط یک ضرب داخلی سریع اجرا می‌کند؛ پس مدل نهایی باید **رگرسیون خطی روی فضای ویژگی‌ها** باشد. چالش شما پاک‌سازی داده و کشف ویژگی‌هایی است که رابطه‌های پنهان را برای مدل خطی آشکار کنند.

### فایل‌ها و ستون‌ها

| فایل | محتوا |
|---|---|
| `data/train/train.csv` | ۱۴۰۰ ردیف برچسب‌دار |
| `data/validation/validation.csv` | ۳۰۰ ردیف برچسب‌دار |
| `data/test/test.csv` | ۳۰۰ ردیف بدون هدف |

ورودی‌ها: `temp`, `ambient_temp`, `pressure`, `rpm`, `hours_active`, `vibration`, `grid_frequency`. هدف `power_output` و شناسهٔ ردیف `id` است. از `id` به‌عنوان ویژگی استفاده نکنید. داده‌ها مقادیر گمشده و مقادیر رشته‌ایِ دارای واحد دارند. آماره‌های پاک‌سازی را فقط از train بیاموزید و همان‌ها را برای validation و test به‌کار ببرید.

### مسیر حل

داده‌ها را پاک‌سازی کنید؛ با نمودار پراکندگی و روند، رابطهٔ تک‌ویژگی‌ها با هدف را ببینید؛ سپس اختلاف دو ویژگی را بررسی کنید. در داده دست‌کم دو رابطهٔ غیرخطی پنهان است: یکی تابعی از یک ویژگی و دیگری تابعی از اختلاف دو ویژگی. آن‌ها را به ویژگی تبدیل کنید و رگرسیون خطی را با NumPy از صفر آموزش دهید.

### مدل فیزیکی و محدودیت‌ها

توان توربین از شرایط حرارتی، فشار، بار مکانیکی و محیط اثر می‌گیرد. مدل نهایی باید خطی در ضرایب باشد؛ تبدیل ورودی‌ها به ویژگی‌های تازه مجاز است. پیاده‌سازی رگرسیون پایه با NumPy الزامی است. کتابخانه‌های مجاز: Python استاندارد، `numpy`, `pandas`, `matplotlib`, `seaborn`, `scipy`. استفاده از `sklearn` یا مدل غیرخطی برای مدل نهایی مجاز نیست.

### معیار و نمره‌دهی

معیار **RMSE** بر حسب مگاوات است. اگر `RMSE > 55` نمره صفر و اگر `RMSE <= 5.5` نمره ۲۰ است؛ در غیر این صورت `min(19, round(5 + 15 * (55 - RMSE) / (55 - 5.5)))`.

## ۴. اکتشاف بصری

نمودارهای تک‌ویژگی و اختلاف دو ویژگی را بررسی کنید. ابتدا ستون‌های عددی را مرور کنید و سپس اختلاف‌های معنادار را رسم کنید.

</div>

```python
def plot_single_feature_with_trend(df, feature, target="power_output", bins=24):
    """Show a scatter plot and binned-median trend using NumPy and Matplotlib."""
    x = pd.to_numeric(df[feature], errors="coerce").to_numpy(float)
    y = pd.to_numeric(df[target], errors="coerce").to_numpy(float)
    valid = np.isfinite(x) & np.isfinite(y)
    x, y = x[valid], y[valid]
    edges = np.linspace(x.min(), x.max(), bins + 1)
    bucket = np.digitize(x, edges[1:-1])
    centers, medians = [], []
    for b in range(bins):
        mask = bucket == b
        if mask.any():
            centers.append(float(np.median(x[mask])))
            medians.append(float(np.median(y[mask])))
    order = np.argsort(centers)
    plt.figure(figsize=(7, 4))
    plt.scatter(x, y, s=12, alpha=0.25)
    plt.plot(np.asarray(centers)[order], np.asarray(medians)[order], color="crimson", marker="o", linewidth=2)
    plt.xlabel(feature); plt.ylabel(target); plt.title(f"{feature} vs {target}")
    plt.grid(alpha=0.25); plt.show()

def plot_difference_with_trend(df, feature_a, feature_b, target="power_output", bins=24):
    """Show the target trend against the difference of two sensor features."""
    temp = df.copy()
    temp["feature_difference"] = temp[feature_a] - temp[feature_b]
    plot_single_feature_with_trend(temp, "feature_difference", target, bins)

# TODO: Clean training data and inspect individual features and selected differences.
```

## ۶. آموزش و پیش‌بینی

مدل و آماره‌های پاک‌سازی لازم را از `train_model` برگردانید. در پیش‌بینی، همان تبدیل‌ها و آماره‌ها را روی query به‌کار ببرید.

</div>

```python
def train_model(frame):
    clean,stats=preprocess_data(frame)
    X=generate_discovered_features(clean); y=clean["power_output"].to_numpy(float)
    model=ScratchLinearRegression().fit(X,y)
    return {"stats":stats,"model":model}
def predict_with_model(state,query):
    clean,_=preprocess_data(query,stats=state["stats"])
    return state["model"].predict(generate_discovered_features(clean))
```

## ۷. ارزیابی محلی

مدل را روی train برازش و RMSE را روی validation گزارش کنید. از validation برای برازش یا محاسبهٔ آماره‌ها استفاده نکنید.

</div>

```python
def rmse(y,p): return float(np.sqrt(np.mean((np.asarray(y)-np.asarray(p))**2)))
model=train_model(train_df)
val_pred=predict_with_model(model,validation_df.drop(columns=["id","power_output"]))
val_rmse=rmse(validation_df["power_output"],val_pred)
print(f"Validation RMSE: {val_rmse:.3f} MW")
```

## ۸. ساخت فایل ارسال

پس از ارزیابی، روی train و validation بازآموزی کنید و روی test پیش‌بینی بگیرید.

```python
# Refit the final model on all labeled rows.
final_train=pd.concat([train_df,validation_df],ignore_index=True)
final_model=train_model(final_train)
predictions=np.asarray(predict_with_model(final_model,test_df.drop(columns=["id"])),dtype=float).reshape(-1)
assert len(predictions)==len(test_df) and np.isfinite(predictions).all()
submission=pd.DataFrame({"id":test_df["id"].to_numpy(),"prediction":predictions})
assert submission.columns.tolist()==["id","prediction"] and submission.id.is_unique
submission.to_csv("submission.csv",index=False)
print(f"Created submission.csv with {len(submission)} rows.")
display(submission.head())
```
